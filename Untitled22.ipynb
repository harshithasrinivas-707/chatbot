```python
import nltk
import random
import string

# Download required NLTK packages
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords

# Initialize NLP tools
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Chatbot response database
responses = {
    "hello": "Hello! Welcome to our chatbot.",
    "hi": "Hi! How can I help you?",
    "hey": "Hey! Nice to meet you.",
    "name": "I am an AI chatbot created using Python.",
    "how": "I am doing great! How are you?",
    "good": "That's great to hear!",
    "bad": "Oh! I hope your day gets better.",
    "help": "Sure! Tell me what you need help with.",
    "study": "Studying regularly helps you achieve your goals.",
    "college": "College life is a wonderful learning experience.",
    "python": "Python is a popular programming language.",
    "ai": "AI stands for Artificial Intelligence.",
    "thank": "You're welcome! Happy to help.",
    "bye": "Goodbye! Have a nice day!"
}

# Text preprocessing function
def preprocess_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Tokenization
    tokens = word_tokenize(text)

    # Remove punctuation
    tokens = [
        word for word in tokens
        if word not in string.punctuation
    ]

    # Stopword removal
    filtered_words = [
        word for word in tokens
        if word not in stop_words
    ]

    # Stemming
    stemmed_words = [
        stemmer.stem(word) for word in filtered_words
    ]

    # Lemmatization
    lemmatized_words = [
        lemmatizer.lemmatize(word) for word in filtered_words
    ]

    print("\n--- NLP Processing ---")
    print("Tokens:", tokens)
    print("After Stopword Removal:", filtered_words)
    print("After Stemming:", stemmed_words)
    print("After Lemmatization:", lemmatized_words)

    return filtered_words


# Chatbot response function
def chatbot_response(user_input):

    words = preprocess_text(user_input)

    # Find matching keywords
    for word in words:

        if word in responses:
            return responses[word]

        # Check stemmed words
        stemmed_word = stemmer.stem(word)

        for key in responses:
            if stemmer.stem(key) == stemmed_word:
                return responses[key]

    return "Sorry, I didn't understand. Can you ask something else?"


# Main chatbot program
def start_chatbot():

    print("\n====================================")
    print(" WELCOME TO AI CHATBOT")
    print("====================================")
    print("You can ask me about Python, AI, college and more.")
    print("Type 'bye' to exit.")

    while True:

        user_input = input("\nYou: ").strip()

        if not user_input:
            print("Chatbot: Please enter something!")
            continue

        if user_input.lower() == "bye":
            print("Chatbot: Goodbye! Have a nice day!")
            break

        answer = chatbot_response(user_input)

        print("\nChatbot:", answer)


# Run chatbot
if __name__ == "__main__":
    start_chatbot()
```
